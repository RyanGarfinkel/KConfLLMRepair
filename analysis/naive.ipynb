```python
import pandas as pd
import json
import glob
import os

path = f'{os.getcwd()}/../workspace/samples/x86-naive/results.json'

rows = []

with open(path, 'r') as f:
    data = json.load(f)

n = data['summary']['n']
successes = data['summary']['successes']
maitenance = data['summary']['maintenance']
build_failures = data['summary']['build_failures']
boot_failures = data['summary']['boot_failures']
truncated = data['summary']['truncated']
avg_duration = data['summary']['avg_duration']

samples = []
for sample in data['samples']:
    samples.append({
        'sample': sample['sample'],
        'boot-status': sample['boot_status'],
        'build-error': sample['build_error'],
        'truncated': sample['truncated'],
        'finish_reason': sample['finish_reason'],
        'input-tokens': sample['token_usage']['input_tokens'],
        'output-tokens': sample['token_usage']['output_tokens'],
        'total-tokens': sample['token_usage']['total_tokens'],
        'duration': sample['duration'] / 60,
    })

df = pd.DataFrame(samples)

print(df['sample'].nunique())

df.head()
```

```python
print('Token Usage:')
print(df[['input-tokens', 'output-tokens', 'total-tokens']].describe())

print('\n\nDuration:')
print(df['duration'].describe())
```
